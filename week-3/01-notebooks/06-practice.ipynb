## Part 2: New Information – Introducing While Loops

### 1. The Shape of a While Loop

```
while condition:
    # code block to execute
    # update something so that eventually the condition becomes False
```

### 2. Counting with a While Loop

````python
# Example 1: Using a while loop with a counter
counter = 1

while counter <= 5:
    print('Counter is:', counter)
    counter = counter + 1  # move the counter forward, or this never ends
````

Adding to a variable is so common that most languages offer a shorthand for it. `counter += 1` means exactly the same thing as `counter = counter + 1`:

````python
counter = 1

while counter <= 5:
    print('Counter is:', counter)
    counter += 1  # shorthand for: counter = counter + 1
````

### 3. When the Condition Starts False

````python
# Example 2: A while loop with a condition that's initially false
counter = 10

while counter < 5:
    print('This will not print')
    counter += 1

print('Loop ended because the condition was false from the start')
````

### 4. Counting Down

````python
# Example 3: A simple countdown using a while loop
count = 3

while count > 0:
    print(count)
    count -= 1  # shorthand for: count = count - 1

print('Blastoff!')
````

### 5. Walking Through a List

````python
colors = ['red', 'blue', 'yellow']
print('There are', len(colors), 'colors in the list')
````

Now the loop. The index starts at `0` and stops before `len(colors)`, because the last valid index is always one less than the length:

````python
colors = ['red', 'blue', 'yellow']
index = 0

while index < len(colors):
    print('Color at position', index, 'is', colors[index])
    index += 1
````

The same pattern does real work if we accumulate a result as we go. Here we total up a list of numbers:

````python
prices = [10, 25, 8]
index = 0
total = 0

while index < len(prices):
    total = total + prices[index]
    index += 1

print('Total:', total)
````

### 6. When Code Breaks

This one does stop. Predict how many times it prints before you run it:

````python
value = 20

while value > 1:
    print(value)
    value = value / 2

print('Done')
````
