### 4. Boolean Values and Boolean Expressions

````python
4 > 3 # is this True or False?
````

````python
# the == operator is used to compare objects
# i.e., are the objects on either side of the == the same?
'Tailor' == 'tailor'
````

````python
3 * 9 == 27 # True or False?
````

The `!=` operator asks the opposite question: are these two things **different**?

````python
print(10 != 10)
print('Tailor' != 'tailor')
````

Let's try some Boolean expressions with variables in the mix...

````python
name == 'Margaret Hamilton'
````

````python
year > 2000 # 21st century? True or False?
````

A Boolean value can be stored in a variable, just like a number or a string:

````python
is_21st_century = year > 2000
print(is_21st_century)
print(type(is_21st_century))
````

### 5. From Algorithm to Code

#### Worked example: convert a Fahrenheit temperature to Celsius

**Step 1: the algorithm** (a non-programmer could follow this)

```
1. Start with a temperature in Fahrenheit
2. Subtract 32 from it
3. Multiply that result by 5/9
4. Report the answer in Celsius
```

**Step 2: the pseudocode** (structure appears, language details don't)

```
set fahrenheit to 72
set celsius to (fahrenheit - 32) times 5/9
display celsius
```

**Step 3: the code.** Each line is commented with the pseudocode step it came from:

````python
fahrenheit = 72                          # set fahrenheit to 72
celsius = (fahrenheit - 32) * 5 / 9      # set celsius to (fahrenheit - 32) times 5/9
print('Celsius:', celsius)               # display celsius
````
