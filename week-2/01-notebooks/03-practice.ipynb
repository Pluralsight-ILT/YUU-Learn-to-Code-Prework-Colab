## Part 2: New Information – Operators and Conditional Logic

### 1. Operator Precedence (Order of Operations)

```python
result = 3 + 4 * 2  # Multiplication happens first
print(result)
```

You can use parentheses `()` to control precedence explicitly, and you should whenever the order isn't obvious at a glance:

```python
result = (3 + 4) * 2
print(result)
```

Precedence matters most when comparisons and arithmetic mix. The arithmetic finishes first, then the comparison runs on the result:

```python
print(2 + 3 > 4)      # the addition happens first, so this is 5 > 4
print((2 + 3) > 4)    # same thing, but now nobody has to guess
```

### 2. Conditionals ("if" statements)

```python
number = 24
```

```python
if number < 24: # this is called an "if statement"
    print('Less than two dozen')
```

```python
if number >= 24: # is number greater than or equal to 24?
    print('Two dozen or more')
```

#### Note that `if` statements can have an optional else part

```python
if number % 2 == 0:
    print(number, 'is even')
else:
    print(number, 'is odd')
```

#### And they can have one or more "else if" parts, which is written `elif` in this language

```python
if number > 0: # is number greater than 0?
    print(number, 'is positive')
elif number < 0: # is number less than 0?
    print(number, 'is negative')
else:
    print('zero')
```

### 3. Logical Operators: `and`, `or`, `not`

**and:**

```python
x = 10
y = 5
print(x > 5 and y < 10)
```

```python
age = 20
is_adult = age >= 18
has_id = True
can_enter = is_adult and has_id  # Both conditions must be True
print(can_enter)
```

**or:**

```python
x = 10
y = 5
print(x < 5 or y < 10)
```

**not:**

```python
x = 10
y = 5
print(not (x > y))
```

Combining them lets one condition express a fairly specific question:

```python
temperature = 68
raining = False

if temperature > 60 and not raining:
    print('Good day for a walk')
else:
    print('Maybe stay in')
```
